```python
import pandas as pd

def rating_to_sentiment(rating):
    if rating <= 2:
        return "Negative"
    elif rating == 3:
        return "Neutral"
    else:
        return "Positive"
    
raw = pd.read_csv(
    "../data/Consumer_Reviews_of_Amazon_Products.csv",
    low_memory=False
)

clean = raw.dropna(subset=["reviews.text", "reviews.rating"])
missing_dropped = len(raw) - len(clean)

duplicates_dropped = clean.duplicated(subset=["reviews.text"]).sum()
clean = clean.drop_duplicates(subset=["reviews.text"]).copy()

clean["sentiment"] = clean["reviews.rating"].apply(rating_to_sentiment)
balance = clean["sentiment"].value_counts(normalize=True).mul(100)
lengths = clean["reviews.text"].str.split().str.len()

print(f"Dataset shape: {raw.shape[0]:,} rows, {raw.shape[1]} columns")
print(
    f"Class balance: Positive {balance.get('Positive', 0):.2f}%, "
    f"Neutral {balance.get('Neutral', 0):.2f}%, "
    f"Negative {balance.get('Negative', 0):.2f}%"
)
print(f"Missing values dropped: {missing_dropped}")
print(f"Duplicate reviews dropped: {duplicates_dropped}")
print(
    f"Review length: min {lengths.min()}, max {lengths.max()}, "
    f"average {lengths.mean():.2f} words"
)
```
